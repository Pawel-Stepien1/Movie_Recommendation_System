# Missing values in data

```python
print(f"Number of NA's in links dataframe:\n{links.isna().sum()}\n")
print(f"Number of NA's in movies dataframe:\n{movies.isna().sum()}\n")
print(f"Number of NA's in ratings dataframe:\n{ratings.isna().sum()}\n")
print(f"Number of NA's in tags dataframe:\n{tags.isna().sum()}")
```

Let's check which rows have missing values in links nad tags

```python
# this code displays rows containing NA's in "links" dataframe
links_nans = links[links.isna().any(axis=1)]
print(links_nans)
```

```python
# this code displays rows containing NA's in "tags" dataframe
tags_nans = tags[tags.isna().any(axis=1)]
print(tags_nans)
```

As we don't have many rows with missing values, we will just delete any data about movies with ID's indicated in above daatframes.

And we will see what happens

```python
# we create list of movies IDs from links_nans
movieId_from_links = links_nans['movieId'].values.tolist()

# we get only unique values from list movieId_from_links
movieId_from_links = list(set(movieId_from_links))
movieId_from_links
```

```python
# we create list of movies IDs from tags_nans
movieId_from_tags = tags_nans['movieId'].values.tolist()

# we get only unique values from list movieId_from_tags
movieId_from_tags = list(set(movieId_from_tags))
movieId_from_tags
```

```python
# we concatenate those two lists
moviesId_to_remove = movieId_from_links + movieId_from_tags
moviesId_to_remove = list(set(moviesId_to_remove))
len(moviesId_to_remove)
```

```python
moviesId_to_remove
```

Okay, now we delete movies with indicated IDs and we'll see what happens

```python
links.drop(links[links['movieId'].isin(moviesId_to_remove)].index, inplace = True)
movies.drop(movies[movies['movieId'].isin(moviesId_to_remove)].index, inplace = True)
ratings.drop(ratings[ratings['movieId'].isin(moviesId_to_remove)].index, inplace = True)
tags.drop(tags[tags['movieId'].isin(moviesId_to_remove)].index, inplace = True)
```
